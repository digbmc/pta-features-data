```python

ss = pd.read_csv('asl-data/all-asl.csv')
source = pd.DataFrame({'category': ['Hard Eight (1996)', 'Boogie Nights (1997)', 'Magnolia (1999)', 'Punch-Drunk Love (2002)', 'There Will Be Blood (2007)', 'The Master (2012)', 'Inherent Vice (2014)', 'Phantom Thread (2017)', 'Licorice Pizza (2021)', 'One Battle After Another (2025)'], 'value': 'seconds'})

categories = ['Hard Eight (1996)', 'Boogie Nights (1997)', 'Magnolia (1999)', 'Punch-Drunk Love (2002)', 'There Will Be Blood (2007)', 'The Master (2012)', 'Inherent Vice (2014)', 'Phantom Thread (2017)', 'Licorice Pizza (2021)', 'One Battle After Another (2025)']
color_mapping = ['#417505', '#F47804', '#52322C', '#BF4983', '#A10215', '#2D5787', '#7ED321', '#B65CE8', '#F5D323', '#4290EB']
```

```python
import pandas as pd
import altair as alt

# --- Assuming categories, color_mapping, and data loading are already defined ---
source = pd.read_csv('asl-data/all-asl.csv')
source2 = pd.read_csv('asl-data/all-asl-data.csv')

selection = alt.selection_point(fields=['Film'], bind='legend')


# 2. Create the first chart (Scatter Plot)
chart1 = alt.Chart(source).mark_circle(size=60).encode(
    x='Shot #',
    y=alt.Y('Shot Length', title="Shot Length (in seconds)"),
    color=alt.Color(
        'Film', 
        scale=alt.Scale(domain=categories, range=color_mapping)),
    tooltip=['Film', 'Shot #', alt.Tooltip('start_time', title='Start Time (in seconds)'), 'Shot Length'],
).add_params(
    selection
).transform_filter(
    selection # <--- This isolate/filters the chart to only the selected category
).properties(
    width='container',
    height=350,
    title="Shot Lengths of All Paul Thomas Anderson Films"
).interactive()

# 3. Create the second chart (ASL Bar Chart)
chart2 = alt.Chart(source2).mark_bar().encode(
    x=alt.X('ASL (in seconds)', title="Seconds"),
    y=alt.Y('Film', axis=alt.Axis(labels=False, title=''), sort=categories),
    tooltip=['ASL (in seconds)'],
    color=alt.Color(
        'Film', 
        scale=alt.Scale(domain=categories, range=color_mapping)),
    # Apply conditional opacity: full opacity (0.8) for selected, faded (0.15) for others
    opacity=alt.condition(
        selection, 
        alt.value(0.8), 
        alt.value(0.15)
    )
).add_params(
    selection
).properties(
    title="ASL",
    height=350,
    width='container'
)

chart1_proportional = chart1.properties(width=800)
chart2_proportional = chart2.properties(width=100)

# 4. Compose and configure charts
composed_chart = chart1_proportional | chart2_proportional

composed_chart = composed_chart.configure_title(
    fontSize=20
).configure_axis(
    titleFontSize=15
).configure_axisX(
    labelFontSize=14
).configure_legend(
    labelFontSize=15,
    titleFontSize=20,
    symbolSize=250,
    labelLimit=0
).properties(
    autosize=alt.AutoSizeParams(type='fit', contains='padding')
)

# Display and save
composed_chart.display()
composed_chart.save('asl-final-scatterplot.html')
```
