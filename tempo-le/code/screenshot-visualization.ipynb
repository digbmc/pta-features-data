```python
import altair as alt
alt.data_transformers.disable_max_rows()
```

```python

ss = pd.read_csv('ss-data/all-in-order-thumb.csv')
source = pd.DataFrame({'category': ['Hard Eight (1996)', 'Boogie Nights (1997)', 'Magnolia (1999)', 'Punch-Drunk Love (2002)', 'There Will Be Blood (2007)', 'The Master (2012)', 'Inherent Vice (2014)', 'Phantom Thread (2017)', 'Licorice Pizza (2021)', 'One Battle After Another (2025)'], 'value': 'seconds'})

categories = ['Hard Eight (1996)', 'Boogie Nights (1997)', 'Magnolia (1999)', 'Punch-Drunk Love (2002)', 'There Will Be Blood (2007)', 'The Master (2012)', 'Inherent Vice (2014)', 'Phantom Thread (2017)', 'Licorice Pizza (2021)', 'One Battle After Another (2025)']
color_mapping = ['#417505', '#F47804', '#52322C', '#BF4983', '#A10215', '#2D5787', '#7ED321', '#B65CE8', '#F5D323', '#4290EB']
```

```python
source = pd.read_csv('ss-data/all-in-order-thumb.csv')

brush = alt.selection_interval(encodings=['x'])

selection = alt.selection_point(fields=['Film'], bind='legend')

base = alt.Chart(source).mark_tick(size=20, thickness=2).encode(
    x='Seconds',
    y=alt.Y('Film', sort=categories, axis=alt.Axis(labels=False)),
    color=alt.Color(
        'Film', 
        scale=alt.Scale(domain=categories, range=color_mapping))
).properties(
    width='container',
    height=400
).add_params(
    selection
).transform_filter(
    selection # <--- This isolate/filters the chart to only the selected category
).properties( # <-- new line for sizing code starts here
    width='container'
)

upper = base.encode(
    alt.X('Seconds', scale=alt.Scale(domain=brush)),
    tooltip=['image', 'Screenshot'],
).properties(
    title="Collection of Paul Thomas Annotated Screenshots"
)

lower = base.mark_tick(
    size=9, 
    thickness=1
).encode(
    # Add the sort parameter here to match the upper chart
    y=alt.Y('Film', sort=categories, axis=alt.Axis(labels=False)) 
).properties(
    height=150, 
).add_params(brush, selection)


vconcat_chart = alt.vconcat(upper, lower).configure_title(
    fontSize=20
).configure_axis(
    titleFontSize=15,
    labelFontSize=14, # Removed the 0 override; explicitly size regular labels
).configure_axisY(
    labelFontSize=0 # Hides Y axis labels cleanly without globally breaking everything
).configure_axisX(
    labelFontSize=14
).configure_legend(
    labelFontSize=15, 
    titleFontSize=20, 
    symbolSize=250,
    labelLimit=0
).configure(
    autosize=alt.AutoSizeParams(type='fit-x', contains='padding')
)        

#vconcat_chart.show()
#vconcat_chart.save('finished-ss-graph.html')
```
